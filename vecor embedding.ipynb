# Import trained model

```python
import gensim.downloader as api
model=api.load("word2vec-google-news-300") #download the model and return as object ready to use
```

# Example of a word as vector

```python
word_vectors=model

# Let us look how the vector embedding of a word looks like 

print(word_vectors['laptop']) # accessing vector for the word "laptop"
```

```python
print(word_vectors["dog"].shape)
```

# Similar words
# King + Woman - Man = ?

```python
# example of using most similar
print(word_vectors.most_similar(positive=['king','woman'],negative=['man'], topn=20))
```

```python
print(word_vectors.similarity('woman','man'))
print(word_vectors.similarity('spiderman','man'))
print(word_vectors.similarity('king','man'))
print(word_vectors.similarity('woman','queen'))
print(word_vectors.similarity('woman','queen'))
print(word_vectors.similarity('water','paper'))
```

# most similar words

```python
print(word_vectors.most_similar('car',topn=10))
```

# now lets see the vector similarity


```python
import numpy as np

# words to compare
word1='man'
word2='woman'

word3='semiconductor'
word4='human'

word5='nephew'
word6='niece'

# calculate the vector differnce

vector_difference1=word_vectors[word1]-word_vectors[word2]
vector_difference2=word_vectors[word3]-word_vectors[word4]
vector_difference3=word_vectors[word5]-word_vectors[word6]

# calculate the magnitude of vector difference
magnitude_of_differnce1=np.linalg.norm(vector_difference1)
magnitude_of_differnce2=np.linalg.norm(vector_difference2)
magnitude_of_differnce3=np.linalg.norm(vector_difference3)

# Print the  magnitude difference
print("'{} and '{}' : {:.2f}".format(word1,word2,magnitude_of_differnce1))
print("'{} and '{}' : {:.2f}".format(word3,word4,magnitude_of_differnce2))
print("'{} and '{}' : {:.2f}".format(word5,word6,magnitude_of_differnce3))
```
